# CS5224 Data Curation — Stage 1A

**Goal:** Load and inspect the three retailer Excel files, confirm their source schemas, map them into one common **raw** schema, and verify the total source row count.

This notebook deliberately does **not** normalize brands, prices, stock values, specifications, categories, or match products across retailers yet. Those are later Stage 1 steps.

Expected source row counts:

- RedMart: **97,779**
- FairPrice: **69,402**
- Cold Storage: **10,457**
- Total: **177,638**

The raw `.xlsx` files are treated as read-only evidence.


```python
from pathlib import Path
import io
import re
import zipfile
import xml.etree.ElementTree as ET

# Put the notebook and the 3 Excel files in the same folder.
# In the ChatGPT environment, the files are in /mnt/data, so the fallback handles that.
DATA_DIR = Path.cwd()

FILENAMES = {
    "redmart": "redmart_all_products_final.xlsx",
    "fairprice": "fairprice_all_products_final.xlsx",
    "coldstorage": "coldstorage_all_products_final.xlsx",
}

if not all((DATA_DIR / name).exists() for name in FILENAMES.values()):
    fallback = Path("/mnt/data")
    if all((fallback / name).exists() for name in FILENAMES.values()):
        DATA_DIR = fallback

FILES = {retailer: DATA_DIR / filename for retailer, filename in FILENAMES.items()}

for retailer, path in FILES.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing file for {retailer}: {path}")

print("Data directory:", DATA_DIR)
for retailer, path in FILES.items():
    print(f"{retailer:12s} -> {path.name}")
```

## Helper functions

The code below reads only the Excel workbook structure and cell values needed for Stage 1A. It does not modify the source workbooks.


```python
MAIN_NS = "http://schemas.openxmlformats.org/spreadsheetml/2006/main"
REL_NS = "http://schemas.openxmlformats.org/officeDocument/2006/relationships"
PKG_REL_NS = "http://schemas.openxmlformats.org/package/2006/relationships"

NS = {"m": MAIN_NS, "r": REL_NS}
PKG_NS = {"p": PKG_REL_NS}


def _shared_strings(zf):
    try:
        root = ET.fromstring(zf.read("xl/sharedStrings.xml"))
    except KeyError:
        return []

    strings = []
    for si in root.findall("m:si", {"m": MAIN_NS}):
        text = "".join((t.text or "") for t in si.iter(f"{{{MAIN_NS}}}t"))
        strings.append(text)
    return strings


def _cell_value(cell, shared):
    cell_type = cell.attrib.get("t")

    if cell_type == "inlineStr":
        inline = cell.find("m:is", {"m": MAIN_NS})
        if inline is None:
            return None
        return "".join((t.text or "") for t in inline.iter(f"{{{MAIN_NS}}}t"))

    value_node = cell.find("m:v", {"m": MAIN_NS})
    if value_node is None:
        return None

    value = value_node.text
    if cell_type == "s":
        return shared[int(value)]
    return value


def _column_index(cell_reference):
    # Example: "A1" -> 0, "L52" -> 11
    letters = re.match(r"[A-Z]+", cell_reference).group()
    index = 0
    for ch in letters:
        index = index * 26 + (ord(ch) - 64)
    return index - 1


def workbook_info(path):
    with zipfile.ZipFile(path) as zf:
        workbook = ET.fromstring(zf.read("xl/workbook.xml"))
        rels = ET.fromstring(zf.read("xl/_rels/workbook.xml.rels"))

        rel_map = {
            rel.attrib["Id"]: rel.attrib["Target"]
            for rel in rels.findall("p:Relationship", PKG_NS)
        }

        sheets = workbook.find("m:sheets", NS)
        if sheets is None or len(sheets) == 0:
            raise ValueError(f"No worksheet found in {path.name}")

        # These files each contain one worksheet.
        sheet = list(sheets)[0]
        sheet_name = sheet.attrib["name"]
        rid = sheet.attrib[f"{{{REL_NS}}}id"]
        target = rel_map[rid]

        sheet_xml = target.lstrip("/") if target.startswith("/") else "xl/" + target
        return sheet_name, sheet_xml


def get_headers_and_row_count(path):
    sheet_name, sheet_xml = workbook_info(path)

    with zipfile.ZipFile(path) as zf:
        shared = _shared_strings(zf)

        # Read the header row.
        root = ET.fromstring(zf.read(sheet_xml))
        header_row = root.find('.//m:sheetData/m:row[@r="1"]', {"m": MAIN_NS})
        if header_row is None:
            raise ValueError(f"Header row not found in {path.name}")

        header_values = []
        for cell in header_row.findall("m:c", {"m": MAIN_NS}):
            idx = _column_index(cell.attrib["r"])
            while len(header_values) < idx:
                header_values.append(None)
            header_values.append(_cell_value(cell, shared))

        # Count actual worksheet row elements, excluding header row.
        row_count = 0
        with zf.open(sheet_xml) as sheet_stream:
            for event, elem in ET.iterparse(sheet_stream, events=("end",)):
                if elem.tag == f"{{{MAIN_NS}}}row":
                    row_number = int(elem.attrib.get("r", "0"))
                    if row_number >= 2:
                        row_count += 1
                    elem.clear()

    return {
        "sheet_name": sheet_name,
        "headers": header_values,
        "data_rows": row_count,
    }


def get_first_data_row(path):
    sheet_name, sheet_xml = workbook_info(path)

    with zipfile.ZipFile(path) as zf:
        shared = _shared_strings(zf)
        root = ET.fromstring(zf.read(sheet_xml))

        header_row = root.find('.//m:sheetData/m:row[@r="1"]', {"m": MAIN_NS})
        data_row = root.find('.//m:sheetData/m:row[@r="2"]', {"m": MAIN_NS})

        headers = []
        for cell in header_row.findall("m:c", {"m": MAIN_NS}):
            idx = _column_index(cell.attrib["r"])
            while len(headers) < idx:
                headers.append(None)
            headers.append(_cell_value(cell, shared))

        values = [None] * len(headers)
        for cell in data_row.findall("m:c", {"m": MAIN_NS}):
            idx = _column_index(cell.attrib["r"])
            if idx < len(values):
                values[idx] = _cell_value(cell, shared)

    return dict(zip(headers, values))
```

## 1. Inspect the three source schemas and count the rows


```python
source_info = {}

for retailer, path in FILES.items():
    info = get_headers_and_row_count(path)
    source_info[retailer] = info

    print("=" * 80)
    print(retailer.upper())
    print("Sheet:", info["sheet_name"])
    print("Rows :", f'{info["data_rows"]:,}')
    print("Columns:")
    for i, column in enumerate(info["headers"], start=1):
        print(f"  {i:2d}. {column}")
```

## 2. Define the common raw schema

At this stage, values are **not cleaned or changed**. Only the column names are standardized.

For example:

- `Keyword` → `source_category_raw`
- `Product Name` → `product_name_raw`
- `Brand` / `brand` → `brand_raw`
- Cold Storage has no `other comments` column, so `comments_raw` will be `None` when rows are transformed later.


```python
COMMON_RAW_FIELDS = [
    "source_category_raw",
    "product_name_raw",
    "product_price_raw",
    "original_price_raw",
    "specification_raw",
    "savings_raw",
    "stock_raw",
    "offer_raw",
    "image_url_raw",
    "product_url_raw",
    "comments_raw",
    "brand_raw",
]

COLUMN_MAPS = {
    "redmart": {
        "Keyword": "source_category_raw",
        "Product Name": "product_name_raw",
        "Product Price": "product_price_raw",
        "Original Price": "original_price_raw",
        "Specification": "specification_raw",
        "Savings": "savings_raw",
        "in-stock": "stock_raw",
        "offer": "offer_raw",
        "Image URL": "image_url_raw",
        "product url": "product_url_raw",
        "other comments": "comments_raw",
        "brand": "brand_raw",
    },
    "fairprice": {
        "Keyword": "source_category_raw",
        "Product Name": "product_name_raw",
        "Product Price": "product_price_raw",
        "Original Price": "original_price_raw",
        "Specification": "specification_raw",
        "Savings": "savings_raw",
        "in-stock": "stock_raw",
        "offer": "offer_raw",
        "Image URL": "image_url_raw",
        "product url": "product_url_raw",
        "other comments": "comments_raw",
        "Brand": "brand_raw",
    },
    "coldstorage": {
        "Keyword": "source_category_raw",
        "Product Name": "product_name_raw",
        "Product Price": "product_price_raw",
        "Original Price": "original_price_raw",
        "Specification": "specification_raw",
        "Savings": "savings_raw",
        "in-stock": "stock_raw",
        "offer": "offer_raw",
        "Image URL": "image_url_raw",
        "product url": "product_url_raw",
        "Brand": "brand_raw",
    },
}

print("Common raw fields:")
for i, field in enumerate(COMMON_RAW_FIELDS, start=1):
    print(f"{i:2d}. {field}")
```

## 3. Validate that every source column is accounted for

This protects us from silently ignoring a source column because of spelling/capitalization differences.


```python
for retailer, info in source_info.items():
    source_headers = set(info["headers"])
    mapped_headers = set(COLUMN_MAPS[retailer])

    unmapped = source_headers - mapped_headers
    unknown_mapping = mapped_headers - source_headers

    print(f"{retailer}:")
    print("  Unmapped source columns :", sorted(unmapped) if unmapped else "None")
    print("  Mapping not in source   :", sorted(unknown_mapping) if unknown_mapping else "None")

    assert not unmapped, f"{retailer} contains unmapped source columns: {unmapped}"
    assert not unknown_mapping, f"{retailer} mapping references missing columns: {unknown_mapping}"
```

## 4. Demonstrate the renamed common raw schema on the first source row

This is only a schema transformation. The values themselves remain raw.


```python
def rename_to_common_raw(retailer, source_row):
    mapping = COLUMN_MAPS[retailer]

    renamed = {field: None for field in COMMON_RAW_FIELDS}

    for source_column, value in source_row.items():
        common_column = mapping[source_column]
        renamed[common_column] = value

    # Cold Storage has no "other comments" field in the raw export.
    # It therefore remains None rather than inventing a value.
    return renamed


sample_rows = {}

for retailer, path in FILES.items():
    source_row = get_first_data_row(path)
    renamed = rename_to_common_raw(retailer, source_row)
    sample_rows[retailer] = renamed

    print("=" * 80)
    print(retailer.upper(), "- first data row after schema rename")
    for field in COMMON_RAW_FIELDS:
        print(f"{field:25s}: {renamed[field]!r}")
```

## 5. Verify the source row counts

The total must equal **177,638** before we move to the next normalization step.


```python
EXPECTED_COUNTS = {
    "redmart": 97_779,
    "fairprice": 69_402,
    "coldstorage": 10_457,
}

for retailer, expected in EXPECTED_COUNTS.items():
    actual = source_info[retailer]["data_rows"]
    assert actual == expected, (
        f"{retailer}: expected {expected:,} rows but found {actual:,}"
    )

total_rows = sum(info["data_rows"] for info in source_info.values())
expected_total = sum(EXPECTED_COUNTS.values())

assert total_rows == expected_total == 177_638

print("Verified row counts")
print("-" * 35)
for retailer in ["redmart", "fairprice", "coldstorage"]:
    print(f"{retailer:12s}: {source_info[retailer]['data_rows']:>8,}")
print("-" * 35)
print(f"{'TOTAL':12s}: {total_rows:>8,}")
print()
print("PASS: Stage 1A source row-count verification succeeded.")
```

## Stage 1A result

At this point we have completed the first Stage 1 task:

- The three raw workbooks remain unchanged.
- Their worksheet names and source schemas have been inspected.
- The `brand` vs `Brand` difference has been resolved at the schema level.
- The missing Cold Storage `other comments` field is represented as `None`.
- All three retailers now map to the same 12-field **common raw schema**.
- The source row counts have been independently checked.
- Total source rows = **177,638**.
